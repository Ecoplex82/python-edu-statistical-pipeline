## Module 2: High-Performance Numerical Analysis (NumPy)

```python
print(">>>>EXTRACTING HIGH-STAKES PERFORMANCE MATTRICS INTO NUMPY MATRIX<<<<<")
metrics_cols = ['math_proficiency_pct', 'reading_proficiency_pct', 'graduation_rate_pct']
metrics_matrix = df[metrics_cols].to_numpy()
```

```python
# 1. VECTORIZED MATRIX NORMALIZAION(Z-score standardization without explicit python loops)
means = np.mean(metrics_matrix, axis= 0)
stds = np.std(metrics_matrix, axis=0)
normalized_metrics = (metrics_matrix - means) / stds
#
```

```python
# 2. linear compositte index calculation via mattrix multiplication
print(">>> Weighs: math proficiency (40%), reading proficiency (40  5), Graduaion Rate(20%)<<<")
weights = np.array([0.40,0.40,0.20])
composite_performance_index = np.dot(normalized_metrics, weights)
```

```python
# 3. boolean masking to identify highy-performing Nodes (>1.5 std Dev above)
print(">>>Boolen Masking<<<")
high_performance_mask = composite_performance_index > 1.5
high_performing_count = np.sum(high_performance_mask)
```

```python
print(">>> MODULE 2 OUTPUT <<<")
print(">>> Module 2: NumPy Vectorized Efficiency Report <<<")
print(f"Metrics Matrix Shape:                  {metrics_matrix.shape}")
print(f"Mean Vector (Math, Reading, Grad Rate): {np.round(means, 2)}")
print(f"Std Vector  (Math, Reading, Grad Rate): {np.round(stds, 2)}")
print(f"High-Performing Institutional Nodes (>1.5 SD): {high_performing_count} institutions")
```

# **Module 4: Quality Diagnostics & Group-Based Imputation**

```python
# 1. Quantifying Missingness Across Indicators
print (">>> Quantifying Missingness Across Indicators <<<")
missing_counts = df.isnull().sum()
missing_pcts = (missing_counts / len(df)) *100

diagnostics_df = pd.DataFrame({
    "Missing Count": missing_counts,
    "NaN Density (%)": missing_pcts
})
```

## *Primary/Middle schools do not administer SAT/ACT or AP/IB exams, resulting in structured NaNs.*

```python
# 2. Smart Group-Based Imputation
print (">>> Smart Group-Based Imputation <<<")
# implementing cluster/group-based median imputation
df['avg_sat_score_imputed'] = df['avg_sat_score'].fillna(
    df.groupby(['level', 'nces_locale'], observed = False)['avg_sat_score'].transform('median')
)
df['ap_ib_participation_imputed'] = df['ap_ib_participation_pct'].fillna(
    df.groupby(['level', 'nces_locale'], observed=False)['ap_ib_participation_pct'].transform('median')
)

remaining_hs_sat_nulls = df[df['level'].isin(['High School', 'K-12 Combined', '4-Year University'])]['avg_sat_score_imputed'].isnull().sum()
```

```python
print (">>> MODULE 4 OUTPUT <<<")
print(">>> Module 4: Data Quality Diagnostics Report <<<")
print("Indicators Demonstrating Missingness:")
print(diagnostics_df[diagnostics_df['Missing Count'] > 0].to_string())
print(f"\nRemaining NaNs for Secondary/Higher Ed SAT Scores Post-Imputation: {remaining_hs_sat_nulls}")
```

## Module 5: Outlier Surgery (Detect, Erase, Heal)**

```python
# logic Guard: Detect Impossible Data (Years_Experience > Age - 18)
print (">>> detecting outliers and impossible Data <<<")
Q1 = df['student_teacher_ratio'].quantile(0.25)
Q3 = df['student_teacher_ratio'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
```

```python
# OUTLIER MASK
print (">>> outlier mask: values outside IQR BOUNDS <<<")

iqr_outlier_mask = (df['student_teacher_ratio'] < lower_bound) | (df['student_teacher_ratio'] > upper_bound)
detected_outliers_count = iqr_outlier_mask.sum()
```

```python
# erasing
print (">>> Erase (NAN conversion via Boolean Indexing <<<)")
df['student_teacher_ratio_healed'] = df['student_teacher_ratio'].copy()
df.loc[iqr_outlier_mask, 'student_teacher_ratio_healed'] = np.nan

```

```python
# HEAL (Calculate district/county level medians)
print (">>> heal (in-place Modification via Districtt/county Median Mapping) <<<")
county_str_medians = df.groupby('county', observed=False)['student_teacher_ratio_healed'].transform('median')
# identifying missing indices creadted during the erasing
nan_indices = df['student_teacher_ratio_healed'].isnull()
# modify dataset in-place using fancy indexing/ boolean Location assignment
df.loc[nan_indices, 'student_teacher_ratio_healed'] = county_str_medians[nan_indices]

remaining_anomalies = df['student_teacher_ratio_healed'].isnull().sum()
```

```python
print (">>> MODULE 5 OUTPUT <<<")

print(">>> Module 5: Outlier Surgery Workflow <<<")
print(f"Student-Teacher Ratio Q1:          {Q1:.2f}")
print(f"Student-Teacher Ratio Q3:          {Q3:.2f}")
print(f"IQR:                               {IQR:.2f}")
print(f"IQR Lower Fence:                   {lower_bound:.2f}")
print(f"IQR Upper Fence:                   {upper_bound:.2f}")
print(f"Detected Outlier Institutions:     {detected_outliers_count}")
print (f" Erased Outliers (Converted to NaN): {df['student_teacher_ratio_healed'].isnull().sum} records")
print(f"Outlier Surgery Status:            Completed successfully. Remaining NaNs = {remaining_anomalies}")
```

# **Module 6: Inferential Statistics & Hypothesis Testing**

Problem Statement 1: Graduation Disparity (Independent T-Test)

- Null Hypothesis ($H_0$): $\mu_{\text{public}} = \mu_{\text{private}}$ (There is no significant difference in graduation rates between Public and Private institutions).

- Alternative Hypothesis ($H_1$): $\mu_{\text{public}} \neq \mu_{\text{private}}$ (There is a statistically significant difference in graduation rates between Public and Private institutions).

Problem Statement 2: Resource Distribution (One-Way ANOVA)
- Null Hypothesis ($H_0$): $\mu_{\text{urban}} = \mu_{\text{suburban}} = \mu_{\text{rural}} = \mu_{\text{town}}$ (Student-teacher ratios are equal across Urban, Suburban, Rural, and Town locales).- Alternative Hypothesis ($H_1$): At least one locale demonstrates a statistically different mean student-teacher ratio.

Problem Statement 3: Stability Analysis (Bootstrap Resampling)
- Requirement: Generate a 95% Confidence Interval for the mean Math Proficiency score using 10,000 resampled iterations.

```python
# Execution 1
print (">>> Graduation Disparity (Independent Welch's T-Test) <<<")
public_grad = df[df['control'] == 'Public']['graduation_rate_pct']
private_grad = df[df['control'].isin(['Private Religious', 'Private For-Profit', 'Private Non-Profit'])]['graduation_rate_pct']

t_stat, p_val_ttest = stats.ttest_ind(public_grad, private_grad, equal_var=False)
```

```python
# Execution 2
print (">>> Resource Distribution (One-Way ANOVA) <<<")
def map_broad_locale(loc_str):
  if 'City' in str(loc_str): return 'Urban'
  if 'Suburban' in str(loc_str): return 'Suburban'
  if 'Town' in str(loc_str): return 'Town'
  if 'Rural' in str(loc_str): return 'Rural'
  return 'Other'

df['broad_locale'] = df['nces_locale'].apply(map_broad_locale)

urban_str = df[df['broad_locale'] == 'Urban']['student_teacher_ratio_healed']
suburban_str = df[df['broad_locale'] == 'Suburban']['student_teacher_ratio_healed']
rural_str = df[df['broad_locale'] == 'Rural']['student_teacher_ratio_healed']
town_str = df[df['broad_locale'] == 'Town']['student_teacher_ratio_healed']

f_stat, p_val_anova = stats.f_oneway(urban_str, suburban_str, rural_str, town_str)
```

```python
# Execution 3
print ("Stability Analysis (Bootstrap Resampling - 10,000 Iterations)")
np.random.seed(42)
scores = df['math_proficiency_pct'].to_numpy()
n_iterations = 10000
sample_size = len(scores)
boottstrap_means = np.empty(n_iterations)

for i in range(n_iterations):
  resample = np.random.choice(scores, size=sample_size, replace=True)
  boottstrap_means[i] = np.mean(resample)

  ci_lower = np.percentile(boottstrap_means, 2.5)
  ci_upper = np.percentile(boottstrap_means, 97.5)
```

```python
print (">>> MODULE 6 OUTPUT <<<")
print(">>> Module 6: Inferential Statistics Workflow <<<")

print(">>> Module 6: Inferential Statistics Results <<<")
print(f"Problem 1 (Welch's T-Test): T-Stat = {t_stat:.4f}, p-value = {p_val_ttest:.4e}")
print(f"  Public Mean Grad Rate:   {public_grad.mean():.2f}%")
print(f"  Private Mean Grad Rate:  {private_grad.mean():.2f}%")
print(f"\nProblem 2 (One-Way ANOVA):  F-Stat = {f_stat:.4f}, p-value = {p_val_anova:.4f}")
print(f"  Locale Means (Urban, Suburban, Rural, Town): [{urban_str.mean():.2f}, {suburban_str.mean():.2f}, {rural_str.mean():.2f}, {town_str.mean():.2f}]")
print(f"\nProblem 3 (Bootstrap 10,000 Resamples):")
print(f"  National Mean Math Proficiency: {scores.mean():.2f}%")
print(f"  95% Confidence Interval:        [{ci_lower:.3f}%, {ci_upper:.3f}%]")
```
